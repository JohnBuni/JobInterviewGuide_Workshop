**Why the misses were wrong**

- **Logistic Regression (intercept & slope).** z = −4 + 1.6 × 2.5 = **0**, and σ(0) = **0.50**: the student sits exactly on the decision boundary (x = −w₀ / w₁ = 2.5 h). The slope adds 1.6 to the log-odds per hour, so the odds multiply by e^1.6 ≈ 4.95.
- **Cross-Entropy.** For y = 1 the loss is −ln(p). −ln(0.9) ≈ **0.11** and −ln(0.1) ≈ **2.30**. The penalty is logarithmic, so a confident wrong prediction costs about 22× more than a confident right one.
- **Data Leakage (feature scaling).** KNN needs scaling because income would dominate every distance. Fitting the scaler on the full dataset leaks test-set statistics into training. A `Pipeline` fits the scaler on training data (or each training fold) only.
- **Decision Tree leaf predictions.** A regression leaf predicts the **mean** of its training targets: (10 + 12 + 14 + 20) / 4 = **14**. A classification leaf predicts the majority class, and its class shares become `predict_proba`.
- **Decision Tree overfitting.** One sample per leaf means the tree memorised the training data. Limit growth with `max_depth` or `min_samples_leaf`, chosen by cross-validation. Trees split on thresholds, so scaling changes nothing.
- **Model choice trade-offs.** When regulators require explanations, interpretability is a hard requirement. Recommend the transparent model, state the 0.02 F1 cost openly, document the decision and keep the stronger model as a benchmark so stakeholders decide with full information.
