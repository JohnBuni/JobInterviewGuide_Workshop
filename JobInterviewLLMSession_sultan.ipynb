**Why the misses were wrong**
- **Q4 MSE.** Squared errors are 1 + 0 + 4 = 5; MSE divides by n → **5 / 3 = 1.67**. 5.00 is the *sum* of squared errors (SSE). MAE = 1.00, RMSE = 1.29.
- **Q5 R².** R² = 1 − SSE/SST compares the model with always predicting the mean. On held-out data it **can be negative** — the model is worse than that baseline.
- **Q8 Sigmoid.** z = −4 + 1.6 × 2.5 = **0**, σ(0) = **0.50**, exactly on the boundary. The slope 1.6 is the change in log-odds per hour; the intercept −4 is the log-odds at 0 hours.
- **Q9 Log-loss.** −ln(0.9) ≈ **0.11**, −ln(0.1) ≈ **2.30**. The penalty is logarithmic, so a confident wrong answer costs ~20× more.
- **Q11 Leakage.** Fitting the scaler on the full dataset leaks test-set mean and std into training. A `Pipeline` fits it on training data or folds only.
- **Q12 Leaf prediction.** A regression leaf predicts the **mean** of its training targets: (10 + 12 + 14 + 20) / 4 = **14**. A classification leaf predicts the majority class; class shares give `predict_proba`.
